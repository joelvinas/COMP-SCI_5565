```r
useSeed <- 2148
set.seed(useSeed)
AllElements <- 392

ratioList <- c(5, 7, 9)

print("2) Compute and report the test performance for 3 new ratios")
for (curRatio in ratioList) {
  curRatioPct <- percent(1/curRatio, accuracy = 1)
  SizeOfSample <- AllElements / curRatio
  train <- sample(AllElements, SizeOfSample)
  print("========")
  print(sprintf("Ratio: %s (1/%s)", curRatioPct ,curRatio))

  #Estimated test MSE for:
  #Linear regression fit
  lm.fit <- lm(mpg ~ horsepower, data = Auto, subset = train)
  linearVal <- mean((Auto$mpg - predict(lm.fit, Auto))[-train]^2)
  print(sprintf("Linear: %s", linearVal))

  #Quadratic
  lm.fit2 <- lm(mpg ~ poly(horsepower, 2), data = Auto,
      subset = train)
  quadraticVal <- mean((mpg - predict(lm.fit2, Auto))[-train]^2)
  print(sprintf("Quadratic: %s", quadraticVal))

  #Cubic
  lm.fit3 <- lm(mpg ~ poly(horsepower, 3), data = Auto,
      subset = train)
  cubicVal <- mean((mpg - predict(lm.fit3, Auto))[-train]^2)
  print(sprintf("Cubic: %s",cubicVal))
  print("========")
}
```

```r
print("3) Compute and report the overall performance of Poly for orders 1:8 for the weight feature in the Auto dataset to mpg.")

useSeed <- 2148
AllElements <- 392
SizeOfSample <- AllElements / 2

set.seed(useSeed)
train <- sample(AllElements, SizeOfSample)

```

```r
print("3) Compute and report the overall performance of Poly for orders 1:8 for the weight feature in the Auto dataset to mpg.")
library(boot)

cv.error <- rep(0, 8)
for (i in 1:8) {
  glm.fit <- glm(mpg ~ poly(weight, i), data = Auto)
  cv.error[i] <- cv.glm(Auto, glm.fit)$delta[1]
}
cv.error
```

```r
print("Using acceleration, perform both a 5 fold and 10 fold k-fold cross validation.")
useSeed <- 2148
set.seed(useSeed)
totalPoly <- 8
kFolds <- 5

print("===  5-fold cross validation ===")
cv.error.totalPoly <- rep(0, totalPoly) #Initialize the vector
for (i in 1:totalPoly) {
  glm.fit <- glm(mpg ~ poly(acceleration, i), data = Auto)
  cv.error.totalPoly[i] <- cv.glm(Auto, glm.fit, K = kFolds)$delta[1]
}
cv.error.totalPoly

kFolds <- 10
print("=== 10-fold cross validation ===")
cv.error.totalPoly <- rep(0, totalPoly) #Initialize the vector
for (i in 1:totalPoly) {
  glm.fit <- glm(mpg ~ poly(acceleration, i), data = Auto)
  cv.error.totalPoly[i] <- cv.glm(Auto, glm.fit, K = kFolds)$delta[1]
}
cv.error.totalPoly
```
